## 4.1 — Why NumPy Exists

Python lists are flexible — they can hold any mix of data types. But that flexibility comes at a **performance cost**.

When doing numerical computation (math on thousands or millions of numbers), Python lists are **slow**. NumPy solves this with arrays that:
- Store only **one data type** (homogeneous)
- Are stored in **contiguous memory** (fast access)
- Support **vectorized operations** (no Python loops needed)

```
Python list: [1, 'hello', 3.14, True]  ← any type, flexible, slow for math
NumPy array: [1, 2, 3, 4, 5]           ← one type, fast, vectorized
```

````python
# Always import NumPy as 'np' — this is the universal convention
import numpy as np

# Quick performance demo: adding 1 to every element
python_list = list(range(1_000_000))
numpy_array = np.arange(1_000_000)

# Python list approach (slow — uses a loop)
%timeit [x + 1 for x in python_list]

# NumPy approach (fast — vectorized, no loop)
%timeit numpy_array + 1
````

## 4.4 — Vectorized Operations & Broadcasting

The superpower of NumPy: **apply operations to entire arrays without writing loops**.

````python
a = np.array([10, 20, 30, 40, 50])  #first array
b = np.array([1,   2,  3,  4,  5])  #second one

print( a + b)
````

````python
a = np.array([10, 20, 30, 40, 50])  #first array
b = np.array([1,   2,  3,  4,  5])

print("a + b =", a + b)     # Element-wise addition
print("a * b =", a * b)     # Element-wise multiplication
print("a ** 2 =", a ** 2)   # Element-wise squaring
print("a / 10 =", a / 10)   # Scalar broadcast — divides EVERY element by 10

# Universal functions (ufuncs)
print("\nsqrt(a):", np.sqrt(a))
print("log(a): ", np.log(a))
print("abs([-1,-2,3]):", np.abs(np.array([-1, -2, 3])))  #absolute number
````

````python
# Aggregation functions
data = np.array([[4, 7, 2],
                 [1, 8, 5]])

print("Sum of all elements:", np.sum(data))
print("-------------------------------------------")

print("Column Sums: ", np.sum(data, axis=0) )

print("Row Sums: ", np.sum(data, axis=1))
````

````python
# Aggregation functions
data = np.array([[4, 7, 2],
                 [1, 8, 5]])

print("Sum of all elements:", np.sum(data))
print("Mean of all elements:", np.mean(data))
print("Max of all elements:", np.max(data))

# The axis parameter is crucial!
# axis=0: operate DOWN the rows → result has one value per COLUMN
# axis=1: operate ACROSS columns → result has one value per ROW
print("\nColumn sums (axis=0):", np.sum(data, axis=0))  # [5, 15, 7]
print("Row sums   (axis=1):", np.sum(data, axis=1))     # [13, 14]

# Cumulative operations
arr = np.array([1, 2, 3, 4, 5])
print("\nCumulative sum:", np.cumsum(arr))  # [1, 3, 6, 10, 15]
````

## 4.5 — Boolean Masking with NumPy

Comparing an array to a value creates a **boolean array** — a mask of True/False values. You can use this mask to **filter** the array.

Data - column os salaries; and we want to filter the data by employees whose salaraies are below 20000 --- when we create a filter of this , that's called masking 
  
#########################

employee_1  == True, 

employee_3 = False

````python
scores = np.array([45, 82, 61, 90, 55, 78, 33, 95])

# passed , should have a score greater than or equal to 60
passed = scores >= 60    # boolean mask 
print(passed)

print("########################################")

#use the boolean mask to filter scores
students_passed = scores[passed]
print(students_passed)




# 1. YOu have to have a condition ---- For a student to pass he/she has to have a score greater that or equalt to 60
# 2. Create a boolean mask ---- passed = scores >= 60
# 3. Use mask to filter the data --- student_passed = scores[passed]
````

````python
scores = np.array([45, 82, 61, 90, 55, 78, 33, 95])

# Create a boolean mask
passed = scores >= 60
print("Scores:", scores)
print("Passed?:", passed)           # [False True True True False True False True]
print("Passing scores:", scores[passed])  # Uses mask to filter — [82 61 90 78 95]

# Combining conditions — MUST use & (AND) and | (OR), not Python's 'and'/'or'!
distinction = (scores >= 75) | (scores <= 100)
print("\nDistinction scores:", scores[distinction])

# np.where: vectorized if/else
# np.where(condition, value_if_true, value_if_false)
grade = np.where(scores >= 60, "Pass", "Fail")
print("\nGrades:", grade)

# Find the index of max/min
print("\nIndex of highest score:", np.argmax(scores), "→ score:", scores[np.argmax(scores)])
print("Index of lowest score: ", np.argmin(scores), "→ score:", scores[np.argmin(scores)])
````

## 5.1 — Column and Row Selection

Understanding `.loc` vs `.iloc` is fundamental. The difference:
- **`.loc`** = **L**abel-based (use the actual name/label)
- **`.iloc`** = **I**nteger position-based (use 0, 1, 2, ... like list indexing)

````python
# --- COLUMN SELECTION ---

# Single column → returns a Series
names_series = employees['name']
print("Type:", type(names_series))  # pandas.core.series.Series

# Single column → returns a DataFrame (note the double brackets [[]])
names_df = employees[['name']]
print("Type:", type(names_df))      # pandas.core.frame.DataFrame

# Multiple columns
subset = employees[['name', 'department', 'salary']]
subset.head()
````

````python
# --- .loc: LABEL-BASED indexing ---
# Syntax: df.loc[rows, columns]
# 'rows' can be: a label, list of labels, boolean array, or slice
# 'columns' can be: a label, list of labels, or slice

# Select row with index label 0, all columns
print("Row 0:\n", employees.loc[0])
print()

# Select rows 0-4, specific columns
employees.loc[0:4, ['name', 'department', 'salary']]
````

````python
# --- .iloc: POSITION-BASED indexing ---
# Syntax: df.iloc[row_positions, col_positions]
# Use integers, just like Python list slicing
# Note: slices EXCLUDE the end, unlike .loc which INCLUDES it!

# First 3 rows, first 4 columns
employees.iloc[0:3, 0:4]
````

## 5.2 — Boolean Filtering

One of the most-used skills in data analysis. Think of it as asking questions about your data.

⚠️ **Critical rule:** Use `&` (AND), `|` (OR), `~` (NOT) — NOT Python's `and`, `or`, `not`!  
⚠️ **Critical rule:** Wrap each condition in parentheses `()` when combining!

````python
# Single condition
senior_staff = employees[employees['age'] > 40]
print(f"Employees over 40: {len(senior_staff)}")
senior_staff[['name', 'age', 'department']]
````

````python
# Multiple conditions with & (AND)
# Find Cape Town Engineering employees
ct_engineering = employees[
    (employees['city'] == 'Cape Town') &
    (employees['department'] == 'Engineering')
]
print("Cape Town Engineering employees:")
ct_engineering[['name', 'city', 'department', 'salary']]
````

````python
# | (OR) condition
management_or_hr = employees[
    (employees['department'] == 'Management') |
    (employees['department'] == 'HR')
]
print(f"Management or HR: {len(management_or_hr)} employees")

# Better way to do OR across many values: .isin()
management_or_hr_v2 = employees[employees['department'].isin(['Management', 'HR'])]
print(f"Using .isin(): {len(management_or_hr_v2)} employees")  # Same result!
````

````python
# ~ (NOT) condition
not_marketing = employees[~(employees['department'] == 'Marketing')]
print(f"Non-marketing departments: {not_marketing['department'].unique()}")

# NOT IN pattern
not_cape_town_jozi = employees[~employees['city'].isin(['Cape Town', 'Johannesburg'])]
print(f"\nEmployees NOT in Cape Town or Johannesburg:")
print(not_cape_town_jozi[['name', 'city']].values)
````

````python
# .between() for range filtering
mid_salary = employees[employees['salary'].between(50000, 70000)]
print(f"Employees earning R50k-R70k: {len(mid_salary)}")

# .query() — string-based filtering (cleaner for complex queries)
# Variables from the environment can be referenced with @
min_salary = 60000
query_result = employees.query("salary > @min_salary and department == 'Engineering'")
print(f"\nEngineers earning > R60k:")
query_result[['name', 'salary', 'department']]
````

## 5.4 — Handling Missing Data

Missing data is unavoidable. There are three main strategies:

1. **Drop** — remove rows or columns with missing values (only if you can afford to lose data)
2. **Fill (Impute)** — replace missing values with a sensible estimate
3. **Flag** — create a new binary column indicating which rows had missing data, then fill

**Theoretical note:** Missing data has three types:
- **MCAR** (Missing Completely At Random) — no pattern; safe to drop
- **MAR** (Missing At Random) — missingness depends on other observed variables; impute carefully
- **MNAR** (Missing Not At Random) — missingness depends on the missing value itself; most problematic

````python
# Detecting missing values
print("Missing value summary:")
missing_summary = pd.DataFrame({
    'Count':   employees.isnull().sum(),
    'Rate (%)': (employees.isnull().mean() * 100).round(1)
})
print(missing_summary[missing_summary['Count'] > 0])
````

````python
# DROPPING
print(f"Original rows: {len(employees)}")

# Drop rows where ANY column is null
dropped_any = employees.dropna()
print(f"After dropna() (any):  {len(dropped_any)}")

# Drop rows where ALL columns are null (almost never happens)
dropped_all = employees.dropna(how='all')
print(f"After dropna() (all):  {len(dropped_all)}")

# Drop rows only if they're missing in SPECIFIC columns
dropped_salary = employees.dropna(subset=['salary'])
print(f"After dropna(subset=['salary']): {len(dropped_salary)}")

# Drop COLUMNS where more than 50% of values are missing
thresh_rows = int(0.5 * len(employees))
dropped_cols = employees.dropna(axis=1, thresh=thresh_rows)
print(f"\nColumns after 50% threshold: {dropped_cols.columns.tolist()}")
````

````python
# FILLING (Imputation)
# Work on a copy so we don't modify the original
emp_clean = employees.copy()

# BEFORE: count missing values
print("Before filling:")
print(emp_clean[['salary', 'age']].isnull().sum())

# Fill numeric with median (robust to outliers — better than mean)
salary_median = emp_clean['salary'].median()
emp_clean['salary'] = emp_clean['salary'].fillna(salary_median)
print(f"\nFilled missing salary with median: R{salary_median:,.0f}")

# Fill numeric with mean (only if distribution is roughly symmetric)
age_mean = emp_clean['age'].mean()
emp_clean['age'] = emp_clean['age'].fillna(round(age_mean))
print(f"Filled missing age with mean: {round(age_mean)}")

# AFTER: verify
print("\nAfter filling:")
print(emp_clean[['salary', 'age']].isnull().sum())
````

````python
# FLAGGING — create a binary indicator BEFORE filling
# This preserves information about which rows were imputed
emp_flagged = employees.copy()

# Create flag columns (1 = was missing, 0 = was present)
emp_flagged['salary_was_missing'] = emp_flagged['salary'].isnull().astype(int)
emp_flagged['age_was_missing']    = emp_flagged['age'].isnull().astype(int)

# Then fill
emp_flagged['salary'] = emp_flagged['salary'].fillna(emp_flagged['salary'].median())
emp_flagged['age']    = emp_flagged['age'].fillna(emp_flagged['age'].mean())

# Check which rows were flagged
flagged_rows = emp_flagged[(emp_flagged['salary_was_missing'] == 1) | (emp_flagged['age_was_missing'] == 1)]
print("Rows that had missing values:")
flagged_rows[['name', 'salary', 'salary_was_missing', 'age', 'age_was_missing']]
````

## 6.2 — GroupBy Operations

The **split-apply-combine** paradigm is the foundation of GroupBy:

1. **Split** the DataFrame into groups based on some key
2. **Apply** a function to each group independently
3. **Combine** the results back into a single output

```
Full DataFrame → [Group A] [Group B] [Group C]  ← SPLIT
                     ↓          ↓          ↓     ← APPLY (e.g., mean)
                   mean_A    mean_B    mean_C
                     ↓                           ← COMBINE
                 [mean_A, mean_B, mean_C]
```

````python
# Basic groupby: average salary per department
dept_avg = employees.groupby('department')['salary'].mean().round(0)
print("Average salary by department:")
print(dept_avg.sort_values(ascending=False))
````

````python
# Group by multiple columns
city_dept = employees.groupby(['department', 'city'])['salary'].mean().round(0)
print("Average salary by department AND city:")
print(city_dept)
````

````python
# .agg() — apply multiple aggregation functions at once
dept_stats = employees.groupby('department').agg(
    headcount     = ('name', 'count'),
    avg_salary    = ('salary', 'mean'),
    min_salary    = ('salary', 'min'),
    max_salary    = ('salary', 'max'),
    total_payroll = ('salary', 'sum'),
    avg_age       = ('age', 'mean')
).round(0)

print("Department statistics:")
dept_stats
````

````python
# .transform() — adds group statistics BACK to each row (same shape as original!)
# Great for creating relative metrics
employees['dept_avg_salary'] = employees.groupby('department')['salary'].transform('mean').round(0)
employees['salary_vs_dept_avg'] = ((employees['salary'] - employees['dept_avg_salary']) / employees['dept_avg_salary'] * 100).round(1)

print("Salary vs department average (% difference):")
employees[['name', 'department', 'salary', 'dept_avg_salary', 'salary_vs_dept_avg']].sort_values('salary_vs_dept_avg', ascending=False)
````

## 6.4 — Melting and Reshaping

**Wide format:** each variable has its own column (e.g., `Jan`, `Feb`, `Mar` as separate columns).  
**Long format:** each observation is a row (e.g., `month` and `sales` as two columns).

Most analysis tools prefer **long format**. Many databases store data in long format.

````python
print("WIDE format (each month = its own column):")
print(sales_wide.head(3))
print(f"Shape: {sales_wide.shape}")

# pd.melt — convert wide to long
sales_long = pd.melt(
    sales_wide,
    id_vars=['emp_id'],           # Columns to keep (identifiers)
    value_vars=['Jan','Feb','Mar','Apr','May','Jun'],  # Columns to 'melt'
    var_name='month',             # Name for the new 'variable' column
    value_name='sales'            # Name for the new 'value' column
)

print("\nLONG format (each row = one employee + one month):")
print(sales_long.head(8))
print(f"Shape: {sales_long.shape}")
````

````python
# Long back to wide: df.pivot()
back_to_wide = sales_long.pivot(
    index='emp_id',
    columns='month',
    values='sales'
)
# Clean up column index name
back_to_wide.columns.name = None
back_to_wide = back_to_wide.reset_index()

print("Back to wide format:")
back_to_wide.head(3)
````

## 6.5 — Combining DataFrames

````python
# pd.concat — stacking DataFrames with the same columns

# Create two batches of new hires
batch1 = employees[['emp_id','name','department','salary']].head(3)
batch2 = pd.DataFrame({
    'emp_id': ['E016', 'E017'],
    'name': ['Uma P', 'Victor N'],
    'department': ['HR', 'Engineering'],
    'salary': [46000, 55000]
})

# Stack rows (axis=0 is default)
combined = pd.concat([batch1, batch2], axis=0, ignore_index=True)
print("Combined (stacked rows):")
print(combined)

# keys parameter — track origin
labeled = pd.concat([batch1, batch2], keys=['existing', 'new_hires'])
print("\nWith origin labels (MultiIndex):")
print(labeled)
````

````python
# Visualise the four join types
print("JOIN TYPES — using sample data:")
print()

left_df  = pd.DataFrame({'id': [1, 2, 3, 4], 'val_L': ['a','b','c','d']})
right_df = pd.DataFrame({'id': [2, 3, 5, 6], 'val_R': ['x','y','z','w']})

print("Left table:",  left_df.values.tolist())
print("Right table:", right_df.values.tolist())
print()

for how in ['inner', 'left', 'right', 'outer']:
    result = pd.merge(left_df, right_df, on='id', how=how)
    print(f"{how.upper():8} join → {len(result)} rows: {result.values.tolist()}")
````

````python
# Best practices for merging

# 1. Check shape before and after — detect unexpected row gain/loss
before = len(employees)
merged = pd.merge(employees, dept_info, on='department', how='left')
after = len(merged)
print(f"Rows before: {before} | Rows after: {after}")
print(f"Row count {'unchanged ✓' if before == after else f'CHANGED — investigate! ({after-before:+d} rows)'}")

# 2. When column names differ, use left_on / right_on
# e.g. pd.merge(df1, df2, left_on='dept_name', right_on='department')

# 3. Handle overlapping non-key column names with suffixes
# e.g. pd.merge(df1, df2, on='id', suffixes=('_employees', '_dept'))

# 4. Validate the relationship type to catch data quality issues
try:
    validated = pd.merge(employees, dept_info, on='department', how='left', validate='m:1')
    print("Validation passed: many employees → one department ✓")
except Exception as e:
    print(f"Validation failed: {e}")
````

## 6.6 — Applying Custom Functions

Sometimes built-in methods aren't enough. `.apply()` lets you run any Python function across a Series or DataFrame.

**Performance note:** Vectorized operations are always faster. Use `.apply()` as a last resort.

````python
# .apply() on a Series (element-wise)
def classify_salary(salary):
    if salary < 50000:
        return 'Below Market'
    elif salary < 75000:
        return 'At Market'
    else:
        return 'Above Market'

employees['market_position'] = employees['salary'].apply(classify_salary)
print("Salary market positions:")
print(employees['market_position'].value_counts())
````

````python
# Lambda functions — inline, anonymous functions for simple operations
employees['salary_k'] = employees['salary'].apply(lambda x: f"R{x/1000:.0f}k")
print("Salary formatted:")
print(employees[['name', 'salary', 'salary_k']].head())

# .apply() on a DataFrame (axis=1 = apply function to each ROW)
# Combine name + department into a display string
employees['display_name'] = employees.apply(
    lambda row: f"{row['name']} ({row['department']})",
    axis=1
)
print("\nDisplay names:")
print(employees['display_name'].values)
````

````python
# Comparing performance: .apply() vs vectorized
big_df = pd.DataFrame({'salary': np.random.randint(30000, 150000, 100_000)})

print("Vectorized (fast):")
%timeit big_df['salary'] * 1.10

print("\n.apply() (slow):")
%timeit big_df['salary'].apply(lambda x: x * 1.10)

print("\n💡 Vectorized is typically 10-100x faster than .apply() for simple operations.")
print("Use .apply() only when you need complex, custom logic that can't be vectorized.")
````

## 🎯 Key Rules to Remember

1. **Always set a random seed** before any random operations: `np.random.seed(42)`
2. **Always run the EDA opening sequence** before any analysis
3. **Use `&`, `|`, `~`** for boolean conditions (NOT `and`, `or`, `not`)
4. **Wrap each condition in parentheses** when combining: `(cond1) & (cond2)`
5. **Use `.loc` to set values**: `df.loc[condition, 'col'] = value`
6. **Use median (not mean)** for imputation when data may have outliers
7. **Flag before you fill** missing values to preserve information
8. **Check `.shape` before and after merging** to detect unexpected row changes
9. **Vectorized operations >> `.apply()`** for performance
10. **Long format is usually better** than wide for analysis and visualization

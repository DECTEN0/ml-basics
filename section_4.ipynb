## Recap: Positional Arguments & Keyword Arguments

```python
def greet(fname, lname):  # we know we need one parameter(name)
    print(f"Hello {fname} {lname}")
```

```python
greet("Ben", "Cook")  #How we've used "Ben" here (argument) ====positional argument
```

```python
greet(lname="Cook", fname="Ben")  #keyword arguments
```

```python
def greet(*names):   # many parameter,  3, 4, 7, 1 
    for name in names:
        print(f"Hello {name}")
```

```python
greet("Ben", "Ian", "Trevor")
```

```python
greet("Ben", "Ian")
```

```python
def my_function(**kwargs):  # Multiple, keyord arguments, name="Ken"    :   name="Ken", age=22 : name="Ken", Age=22, Gender="Male"
    print("Type: ", type(kwargs))
    print("Name: ", kwargs["name"])
    print("Age: ", kwargs["age"])
    print("All Data: ", kwargs)
```

```python
my_function(name = "Ian", age=22, friend = "John", teacher="Ken")
```

## Lambda Functions
A lambda function is a small anonymous function used for short, one-time operations.

```python
# function to calculate the square of a number 5 == 25
def square_5(x):
    calculation = x * x
    return calculation
```

```python
square_5(x=5)  #9ms
```

```python
squre_lambda = lambda x: x **2   #this is the function
```

```python
print(squre_lambda(5))  #2ms
```

```python
square = lambda x: x ** 2
print(square(5))
```

### Using lambda with map(), filter(), and sorted()

```python
function_to_square = lambda x: x**2   #function to vsqure numbers

numbers = [1, 2, 3, 4, 5]  # numbers to square can be a list, set, tuple, dictionaries

squared_numbers = list(map(function_to_square, numbers))

print(squared_numbers)

#output = [1, 4, 9, 16, 25]
```

```python
my_tuple = (1, 2, 3)

list(my_tuple)    
```

Filter function used to extract elements from an iterable (list, tuple, sets, dictionary) that satisfy a specific condition

```python
"""
    Our goal is to get even numbers in a list
"""
numbers = [1, 2, 3, 4, 5]  


#Will return True if even else False
even_lambada = lambda x: x% 2 == 0   #will return true for even numbers

list(filter(even_lambada, numbers))
```

```python
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x**2, numbers))
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print('Squared:', squared)
print('Even:', even_numbers)
```

Sorted function - sort any iterable object (lists, tuples, sets, dictionaries) without modiflying the original data


Return a list

```python
numbers = [42, 4, 15, 8, 23]

my_tuple = (3, 1, 2)


print(sorted(my_tuple))
print(sorted(numbers))
```

### List Comprehension Alternatives

```python
numbers = [42, 4, 15, 8, 23]

squared_numbers = []

for num in numbers:
    squared = num ** 2
    squared_numbers.append(squared)

print(squared_numbers)   #2ms
```

```python
numbers = [42, 4, 15, 8, 23]
squared_numbers_comp = [num ** 2 for num in numbers]
print(squared_numbers_comp)
```

```python
squared = [x**2 for x in numbers]
even_numbers = [x for x in numbers if x % 2 == 0]

print(squared)
print(even_numbers)
```

# 4.4 Exception Handling

Get errors, how do you do you wanna deal with the errors?

```python
2 / 0   
```

```python
try:
    #code you wanna try
    2 / 0
except ZeroDivisionError:
    # do this if try fails
    print("Cannot Divide by Zero")
```

```python
try:
    num = int(input('Enter a number: '))
    print(10 / num)
except ValueError:
    print('Please enter a valid integer.')
except ZeroDivisionError:
    print('Cannot divide by zero.')
else:
    print('No errors occurred.')
finally:
    print('This always runs.')
```

```python
def validate_age(age):
    if age < 0:
        raise ValueError('Age cannot be negative')
    return age

print(validate_age(1))
```

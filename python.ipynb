## Loops

### For Loops

```python
# lists, dictionaries, strings, dataframes, sets

list1 = [100,200,300,400,500,600,700]

for item in list1:
    print(item)
    
```

```python
# lists, dictionaries, strings, dataframes, sets

list1 = [100,200,300,400,500,600,700]

for i in range(len(list1)):
    #get the item at theat index
    item = list1[i]

    #index of the item alongside the item iteself
    print(i, item)
```

```python
list1 = [100,200,300,400,500,600,700]

for i, item in enumerate(list1):
    print(i, item)
```

```python
dict1 = {1:"Ken", 2:"Ian", 3:"Trevor"}

for value in dict1.values():
    print(value)
```

```python
dict1 = {1:"Ken", 2:"Ian", 3:"Trevor"}

for key in dict1.keys():
    print(key)
```

```python
dict1 = {1:"Ken", 2:"Ian", 3:"Trevor"}

for key,value in dict1.items():
    print(key, value)
```

```python
list1 = [100,200,300] #marks
list2 = ["Ken", "Ian", "Trevor"] #names

#ken : 100
#ian : 200
#trevor : 300

for name, marks in zip(list2, list1): #[("ken", 100), ("ian", 200), ("Trevor", 300)]

    print(f"{name} : {marks}")
    
```

```python
#while loop --- infinite loop  --- when you don't know
#you don't know when your loop will end
#you know a stopping condition

x = 0
while True:  #infinite loop
    if x > 9:  #stopping condition
        break
    print(x)
    x += 1  # x = x + 1
```

```python
x = 0
while x < 10:
    print(x)
    x += 1
```

### Basic Function

```python
#create the function
def greet(name: str):
    print(f"Hello {name}")
```

```python
#calling the function
greet(name="Ben")  # name is the Parameter, Ben is the argument
```

```python
greet("Ian")
```

### Improved Functions

`name` in the function below is called a `parameter`. A value that changes. 

An `argument` is the actual value that is sent to the function while it's called. eg for the above examples the arguments are `Ben` and `Ian`.

```python
#create the function
def greet(fname, lname):
    print(f"Hello {fname} {lname}")
```

```python
#call --- will follow the order of the parameters in the function
greet("Ben", "Ten")
```

```python
greet(lname="Ten",fname="Ben")
```

```python
greet("Ten", "Ben")
```

### Default Parameters

```python
def greet(fname="Ben", lname="Ten"):  #default value 
    print(f"Hello {fname} {lname}")
```

```python
greet("Ben")  #lastname will take default
```

```python
greet("Ben","Cook")   
```

### Python *args and **kwargs

`*args` = arguments
`**kwargs` = key word arguments

May not know how many parameters/ arguments that will be passed into your function.

`*args` and `**kwargs` allows the function to accept an unkow number of arguments.

```python
def greet(*args):
    print("Type:", type(args))

    print("First arg:", args[0])

    print("Second arg: ", args[1])
    
    print(args)

```

Positional arguments- depend on the postion of the arguments as you call your function.

```python
greet("Ian", "and", "John")
```

```python
def greet(*names):
    for name in names:
        print(f"Hello {name}")
        
```

```python
greet("Ian", "Ben", "Cook", "Trevor", "Ken")
```

### Arbitrary Keyword Arguments - **kwargs

Allow a function to accept any number of keyword argumments.

```python
def my_function(**kwargs):
    print("Type: ", type(kwargs))
    print("Name: ", kwargs["name"])
    print("Age: ", kwargs["age"])
    print("All Data: ", kwargs)
```

```python
my_function(name = "Ian", age=22, friend = "John", teacher="Ken")
```

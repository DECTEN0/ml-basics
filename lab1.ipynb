## Training a basic perceptron model

```python
#Imports
import pandas as pd
from sklearn.neural_network import MLPClassifier, MLPRegressor
```

### Load data

```python
#Load the data
my_cols = ['x1', 'x2', 'target']    #column names for the data
df = pd.read_csv('Percept1.csv', sep=';', header=None, names=my_cols)

#Show the first few rows of the data
df.head()
```

```python
#observe data xtics
print(f"Unique values in 'target' column: {df['target'].unique()}")

#print shape of the data
print(f"Shape of the data: {df.shape}")
```

```python
#check for missing values
print(f"Missing values in each column:\n{df.isnull().sum()}")
```

### Modeling

```python
#neural network classifier
X = df[['x1', 'x2']]  #features
y = df['target']     #target variable
model = MLPClassifier(hidden_layer_sizes=(0,0), max_iter=1000,shuffle=True, random_state=42, verbose=True)
model.fit(X, y)
```

```python
#make predictions
predictions = model.predict(X)
print(f"Predictions: {predictions}")
```

```python
#save the model
import pickle  #file name   

with open('perceptron_model.pkl', 'wb') as f:  #context manager used to store data models
    pickle.dump(model, f)
```

```python
#load the model
with open('perceptron_model.pkl', 'rb') as f:  # load our model 
    loaded_model = pickle.load(f)
```

```python
#see the params of the model
print(f"Model parameters: {loaded_model.get_params()}")
```
